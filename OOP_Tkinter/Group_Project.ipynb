### Project Description: Ride Booking System
Develop a Ride Booking System that allows users to book rides, view available rides, cancel bookings, and manage different types of vehicles. The application will utilize object-oriented programming principles, including inheritance, polymorphism, and encapsulation. Your group will also implement file handling for data persistence and a graphical user interface using Tkinter.

### Objectives:
- To apply object-oriented programming concepts such as classes, inheritance, and polymorphism.
- To learn and implement file handling in Python for saving and retrieving application data.
- To design and develop a graphical user interface using Tkinter.
- To understand and implement basic software engineering principles such as modularity and encapsulation in a practical application.

### System Features:
1. **Vehicle Management**:
   - **Different Vehicle Types**: Implement at least three different types of vehicles (e.g., Car, Van, Bike), each with specific properties such as cost per mile and capacity.
   - **Vehicle Cost Calculation**: Use polymorphism to override the cost calculation method for each vehicle type based on specific criteria (e.g., fuel efficiency, luxury tax).

2. **Booking Features**:
   - **Book a Ride**: Users can choose a vehicle type, specify start and end locations, and book a ride. The system calculates the cost based on the distance and selected vehicle type.
   - **View All Bookings**: Display a list of all current bookings with details such as booking ID, user, vehicle type, start and end locations, distance, and total cost.
   - **Cancel Booking**: Allow users to cancel their bookings. The system should update the list of available bookings accordingly.

3. **File Operations**:
   - **Save to File**: Automatically save all bookings to a file when a booking is added or cancelled.
   - **Load from File**: Load bookings from a file at startup, ensuring that data persists between sessions.

4. **User Interface**:
   - Develop a GUI using Tkinter that includes forms for entering booking details, buttons for submitting or cancelling bookings, and a display area for viewing all bookings.

5. **Group Portfolio**:

### Technical Requirements:
- **Python**: Use Python 3.x.
- **Tkinter**: Use Tkinter for the GUI.
- **OOP Principles**: Properly utilize inheritance, polymorphism, and encapsulation.

### Expected Outcomes:
- **Functioning Application**: A working application that meets all the specified features.
- **Code**: Well-organized and commented source code.
- **Documentation**: A report documenting the design decisions, challenges faced, and a user guide for using the application.
- **Presentation**: A brief presentation explaining the work, showcasing the application functionality and discussing the learning points.

### Evaluation Criteria:
- **Functionality (30%)**: How well does the application work? Does it meet all the feature requirements?
- **Code Quality (30%)**: Is the code well-structured and commented? Are OOP principles effectively applied?
- **User Interface (20%)**: Is the GUI intuitive and responsive?
- **Documentation and Presentation (20%)**: How well is the project documented and presented?








### Project Description: Learning Management System (LMS)
Develop a Learning Management System (LMS) that allows instructors to add courses, students to enroll in courses, manage their enrollments, and access course materials, including PDF lecture files. The system will be designed using object-oriented programming principles and will include file handling for data persistence and PDF management.

### Objectives:
- To apply object-oriented programming concepts including classes, inheritance, and polymorphism.
- To implement file handling for saving, retrieving, and managing PDF files.
- To design and develop a graphical user interface using Tkinter.


### System Features:
1. **Course Management**:
   - **Add Course**: Instructors can create new courses with details such as course name, description, and associated PDF materials.
   - **Edit Course**: Instructors can update course details and change PDF files.
   - **Delete Course**: Instructors can remove courses from the system.

2. **Enrollment Features**:
   - **Enroll in Course**: Students can browse available courses and enroll in them.
   - **View Enrolled Courses**: Students and instructors can view a list of courses in which they are enrolled.
   - **Unenroll from Course**: Students can withdraw from courses.

3. **PDF Management**:
   - **Upload PDF**: Instructors can upload PDF files for each course.
   - **Download PDF**: Students can download and view PDF files associated with their enrolled courses.
 
4. **User Interface**:
   - A GUI that facilitates the management of courses, enrollments, and PDF files. This should include forms for adding/editing courses, lists for viewing available and enrolled courses, and options for uploading and downloading PDF files.

5. **Group Portfolio**

### Technical Requirements:
- **Python**: Use Python 3.x for backend logic.
- **Tkinter**: Use Tkinter for creating the GUI.
- **File Handling**: Implement robust file handling to manage course data and PDF files securely.

### Expected Outcomes:
- **Functioning Application**: The application should function as described, with all features implemented.
- **Code**: The source code should be well-organized, commented, and adhere to OOP principles.
- **Documentation**: Complete documentation covering design decisions, implementation details, and a user guide.
- **Presentation**: Optionally, a presentation that summarizes the project, showcases key features, and discusses challenges and solutions.

### Evaluation Criteria:
- **Functionality (30%)**: How well does the application work? Does it meet all the feature requirements?
- **Code Quality (30%)**: Is the code well-structured and commented? Are OOP principles effectively applied?
- **User Interface (20%)**: Is the GUI intuitive and responsive?
- **Documentation and Presentation (20%)**: How well is the project documented and presented?
